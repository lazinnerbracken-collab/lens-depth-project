```python
import librosa
import numpy as np
import matplotlib.pyplot as plt
import scipy.io.wavfile as wav
import IPython.display as ipd
```

```python
machine = 'ToothBrush'
# machine = 'ExtraTests'

file = 'section_00_source_train_normal_0399_noAttribute'
# file = 'Slider_0010'
```

```python
audio_path = 'D:/uni/TER LD/data/sounds/%s/train/%s.wav'%(machine, file)
audio, sample_rate = librosa.load(audio_path, mono=False, sr=None)

audio1 = audio[0]
audio2 = audio[1]
```

audio1 désigne le micro proche et audio2 le micro éloigné

```python
ipd.Audio(audio1, rate=sample_rate)
```

```python
ipd.Audio(audio2, rate=sample_rate)
```

```python
plt.figure(figsize=(10, 4))
librosa.display.waveshow(audio1, sr=sample_rate, color='darkblue')
plt.title('audio proche')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 4))
librosa.display.waveshow(audio2, sr=sample_rate, color='orange')
plt.title('audio loin')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 4))
plt.plot(audio1, color='darkblue', alpha=0.7, label='proche')
plt.plot(audio2, color='orange', alpha=0.4, label='éloigné')
plt.title('Audios')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
stft = librosa.stft(audio1)
spectrogram = librosa.amplitude_to_db(np.abs(stft))
plt.figure(figsize=(10, 4))
librosa.display.specshow(spectrogram, sr=sample_rate, x_axis='time', y_axis='log', cmap='viridis')
plt.colorbar()
plt.title('Spectrogram proche (dB)')
```

```python
stft = librosa.stft(audio2)
spectrogram = librosa.amplitude_to_db(np.abs(stft))
plt.figure(figsize=(10, 4))
librosa.display.specshow(spectrogram, sr=sample_rate, x_axis='time', y_axis='log', cmap='viridis')
plt.colorbar()
plt.title('Spectrogram éloigné (dB)')
```
